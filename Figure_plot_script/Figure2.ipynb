```python

net_stas_all=pd.read_csv('ts_mut_exp_net_num.txt',sep='\t')
net_stas_all_sort=net_stas_all.sort_values('Tissue-specific mutated genes')
net_stas_all_sort['TSMGs without identified subnetworks']=net_stas_all_sort['Tissue-specific mutated genes']-net_stas_all_sort['Tissue-specific subnetworks']
net_stas_all_sort['TSMG subnetworks']=net_stas_all_sort['Tissue-specific subnetworks']
custom_palette = {"TSMGs without identified subnetworks": "#CAB782", "Tissue-specific subnetworks": "#3677A0"}  # You can customize colors here
plot_cols = ['Tissue-specific subnetworks','TSMGs without identified subnetworks']
colors = [custom_palette[c] for c in plot_cols]
fig,axes=plt.subplots(1,1,figsize=(8.5/2,11/4))
plt.rcParams['axes.linewidth'] = 0.5
axes.spines['top'].set_visible(False)
axes.spines['right'].set_visible(False)
sns.set_style("ticks")
# 绘制双柱状图
sns.set_style("ticks")

# 绘制堆叠柱状图
net_stas_all_sort[plot_cols].plot(
    kind='bar',
    stacked=True,
    color=colors,
    ax=axes,
    width=0.8,
    edgecolor='none'
)
# 设置Y轴标签
axes.set_ylabel('Number',fontsize=8, fontname='Arial')
axes.set_xlabel('')
# 设置标题
axes.set_title('',fontsize=8, fontname='Arial')
axes.set_xticklabels(net_stas_all_sort['Tissue'].drop_duplicates().tolist(),fontsize=8, fontname='Arial')
axes.tick_params(labelrotation=90,axis='x')
# 显示图例
plt.legend(prop={'family': 'Arial',  'size': 8})
plt.savefig(f'tsm_network_ticks.pdf',dpi=300,bbox_inches='tight')
```
